# *Comprehension*

A "*list comprehension*" allows you to build new lists (`list`) by filtering the elements of a collection (sequence) with a single expression. Its traditional form is the following:

```
[expr for val in collection if condition]
```
the above is equivalent to:

```
res = []
for val in collection:
  if condition:
    result.append(expr)
```


For example:

````
list2 = ['Facebook', 'Apple', 'Amazon', 'Netflix']

list2_title = [elem.upper() for elem in list2]
list2_title 
````

Which in its standard form is equivalent to:

````
list2_title2 = []

for elem in list2:
  list2_title2.append(elem.upper())

list2_title2
````

You can even add the `else` conditional, which changes the structure of the *list comprehension* slightly, but the goal is the same:

````
[num if (num%3 == 0) or (num%5 == 0) else None for num in range(16)]
````

Which in its standard form is equivalent to:

````
res2 = []

for num in range(16):
  if (num%3 == 0) or (num%5 == 0):
    res2.append(num)
  else:
    res2.append(None)

res2
````

It can also be used to "flatten" nested lists:

````
nested_list = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

[sub_elem for elem in nested_list for sub_elem in elem]
````

Which in its standard form is equivalent to:

````
flattened_list = []

for elem in nested_list:
  for sub_elem in elem:
    flattened_list.append(sub_elem)

flattened_list
````

To see *comprehension* for dictionaries (`dict`) and tuples (`tuple`), see pages 67 and 68 of the book *Python for Data Analysis: Data Wrangling with Pandas, Numpy, and Ipython* by Wes McKinney.
