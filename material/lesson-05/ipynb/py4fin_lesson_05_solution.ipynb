# Arrays

Objects such as lists (`list`) are very convenient because of their flexibility (they are mutable); however, this kind of structure has a cost: it uses a lot of memory and its performance is poor. Financial and/or scientific applications require high-performance operations on special structures. One of the most important data structures is the *array*. *Arrays* usually organize other objects of the same data type both in rows and in columns.

If we assume that the most relevant data types are `float` and `int` (the idea can be extended to other data types), a one-dimensional `array` would represent, mathematically speaking, a row or column vector (depending on how it is defined). The most common cases are matrices, which would be a two-dimensional `array` $i\times j$ (row and column).

We therefore need a class capable of working efficiently with *arrays*. The solution is the **NumPy** library with its `ndarray` class.

If we want to create the matrix $X$:

$$
X = \begin{pmatrix}
1 & 2 & 3 \\
4 & 5 & 6 \\
7 & 8 & 9
\end{pmatrix}
$$

With lists (`list`):

```
x = list([[1,2,3],[4,5,6],[7,8,9]])
x
```

Where each element nested in the list would be a row of the matrix $X$ and each element of the nested element a column. If we want to see the value at position $2\times 2$:

```
x[1][1]
```

However, Python interprets the variable $X$ as a list and not as an `array`. Using `numpy`

```
import numpy as np

mat_x = np.array(x) # the array is generated
mat_x
```

Now the variable `mat_x` is an *array* and as such it has the following attributes: 

* Dimensions:

```
mat_x.ndim # it has two dimensions
```

* The values of the dimensions:

```
m, n = mat_x.shape # number of rows and columns
m, n
```

* Total number of elements:

```
mat_x.size
```

* New class:

```
type(mat_x)
```

the `range()` version that is *built-in* in Python exists in **NumPy** as `arange()`. 

```
np.arange(1,10).reshape(-1,3)
```

The library documentation can be found at https://numpy.org/doc/stable/reference/.

```
'''1.a'''
num = [3, 0, 2, 2, 0, 2, 0, 1, 1]

list_to_array = np.array(num)
a = list_to_array.reshape((3,3))

# way 1
a.T

# way 2
a.transpose()

'''1.b'''
b = np.linalg.inv(a)

'''1.c'''
a.dot(b)

# to generate an identity matrix 
np.eye(3)

'''1.d'''
mean_a = np.ones((1,3)).dot(a).dot(np.ones((3,1))) / a.size
mean_a[0][0]
```

3. Write a function that multiplies (not the dot product) two matrices.

```
mat_a = np.array([[1,2,3],[4,5,6]])
mat_b = np.array([[1,2,3],[4,5,6]])
```

```
def manual_mult(mat_a, mat_b):
  '''
  multiplies element ij of matrix a with element ij of matrix b 
  ''' 
  m,n = mat_a.shape

  zero_mat = np.zeros((m,n))
  zero_mat.fill(np.nan)

  for i in range(m):
    for j in range(n):
      zero_mat[i][j] = mat_a[i][j] * mat_b[i][j] 

  return(zero_mat)

# we test the function
manual_mult(mat_a, mat_b)
```

# Random Numbers

**NumPy** has its own pseudo-random number generator (Mersenne Twister). The generator is a mathematical function that produces a sequence of "random numbers". It takes a parameter to start the sequence of numbers, called the seed (`seed`). The function is deterministic, that is, with the same seed it will always produce the same sequence of numbers (the choice of the seed does not matter).

1. Generate 10000 values from a standard normal distribution $\sim N(\mu = 0, \sigma = 1)$.

  a. From the values created, generate a matrix $E$ whose dimensions are $100 \times 100$. What is the difference between `reshape()` and `resize()`?

  b. "Flatten" the matrix $E$ and verify that the data are indeed distributed with $\mu = 0$ and $\sigma = 0$.

```
'''1.a'''
np.random.seed(10)

array_norm = np.random.randn(10000)

# reshape
e = array_norm.reshape((100,100)) 

# resize
array_norm.resize((100,10), refcheck=False)
array_norm # in-place, but it must not have been referenced, for example with reshape

'''1.b'''
e_flatten_col = e.flatten(order='C') # flattened by column
e_flatten_row = e.flatten(order='F') # flattened by row

e_flatten_col.mean() # mean
e_flatten_col.std() # standard deviation

array_norm
```

# System of Equations

1. Use matrices to solve the following systems of equations:

  a.
  $$
  \begin{align*}
    a + b + c = 6 \\
  3a - 2b + c = 2 \\
   2a + b - c = 1
  \end{align*}
  $$

  b.
  $$
  \begin{align*}
  3a + 4b − 5c + d = 10 \\
  2a + 2b + 2c − d = 5 \\
   a − b + 5c − 5d = 7 \\
            5a + d = 4
  \end{align*}
  $$

```
'''1.a'''
# the matrices are generated (unknowns + result)
mat1 = np.array([[1, 1, 1], [3, -2, 1], [2, 1, -1]])
mat1_res = np.array([6, 2, 1])

# the system of equations is solved
np.linalg.solve(mat1, mat1_res)

'''2.b'''
# the matrices are generated (unknowns + result)
mat2 = np.array([[3, 4, -5, 1], [2, 2, 2, -1], [1, -1, 5, -5], [5, 0, 0, 1]])
mat2_res = np.array([10, 5, 7, 4])

# the system of equations is solved
np.linalg.solve(mat2, mat2_res)
```
