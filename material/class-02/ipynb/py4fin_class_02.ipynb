## Tuple (`tuple`)

A tuple (`tuple`) is created if you use a comma (,) to separate values when assigning to a single variable:

```
tuple_ = 1, 2, 'Facebook', 'Amazon'
tuple_
```

Just like lists, you can create nested tuples (`tuple`):

```
nested_tuple1 = (1,2), ('Facebook', 'Amazon')
nested_tuple1
```

They can even store lists (`list`):

```
nested_tuple2 = (1,2), ('Facebook', 'Amazon'), ['Apple', 'Netflix']
nested_tuple2 
```

You can only modify an element of the tuple (`tuple`) that allows modification, for example a list (`list`):

```
nested_tuple2[2].append('Google')
nested_tuple2
```

It allows concatenation using the `+` sign:

```
('Facebook', 'Amazon') + ('Apple', 'Netflix') 
```

And even multiplication of elements using `*`:

```
('Facebook', 'Amazon') * 3
```

To force a list (`list`) to become a tuple (`tuple`) it is enough to write:

```
tuple([1, 2, 3, 4])
```

## Dictionary (`dict`)

Dictionaries (`dict`) are mutable data structures, characterized by the use of key-value pairs (*key-value*).

```
dict1 = {'Name' : 'Janet Yellen', 
         'Country' : 'United States',
         'Profession' : ' United States secretary of the treasury',
         'Age' : 74}
        
type(dict1)
```

To access the values associated with a key, it is enough to use `[ ]` and write the name of the key:

```
print(dict1['Name'], dict1['Age']) 
```

It inherits the `.keys()` method to access the names of the keys:

```
dict1.keys() # keys
```

It inherits the `.values()` method to access the values associated with the keys:

```
dict1.values() # values 
```

With `.items()` you access both the key (*key*) and the values (*values*):

```
dict1.items()  # items = keys + values
```

## Set (`set`)


The set (`set`) data structure allows only unique values. To create a set (`set`), you create a list and then apply `set()`: 

```
set1 = set([1, 2, 3, 4, 5, 5])
set1
```

```
set2 = set([3, 4, 5, 6, 9, 9, 7])
set2
```

We can apply set operations:

* All the items that are in `set1` and `set2`:

```
# union
set1.union(set2)
```

* Items that are in both `set1` and `set2`:

```
# intersection
set1.intersection(set2)
```

* Items that are in `set1` but not in `set2`:

```
set1.difference(set2)
```

* Items that are in `set2` but not in `set1`:

```
set2.difference(set1)
```

* Items that are in `set1` or `set2` but not in both:




```
set1.symmetric_difference(set2) 
```
