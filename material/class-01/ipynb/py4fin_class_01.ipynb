# Variables

A variable in Python is a space in your computer's RAM where you can store results and objects.
To create a variable we use an "assignment", that is, an expression with an equality. What is to the left of the equals sign will be the name of the variable where the value is stored, and what is to the right of the equals sign is the value to be stored (*Aplicaciones de Manejo de Datos, 2020*).

For example, let's create the variable `phone_number` and store the value `12345678`:

```python
phone_number = 12345678 # the symbol = is the assignment operator
phone_number
```

In python it is not necessary to end a line of code with ; (semicolons), but they do serve a purpose:

```python
a = 1; b = 2; c = 3
```

You can also assign values to several variables in a single line of code:

```python
a, b, c = 1, 2, 3
```
