Developed by William Sharpe (1964), the CAPM (*Capital Asset Pricing Model*) starts from the idea that every investment carries two types of risk: systematic (market risk that cannot be diversified away) and unsystematic (the component of a stock's return that is not correlated with market movements). portfolio diversification can remove unsystematic risk; however, this is not the case for systematic risk. 

The CAPM is a way of measuring systematic risk. William Sharpe finds that the return of a stock or portfolio is:

<br>
$$
R_{i,t} = R_f + \beta_i (R_{m,t} - R_f)
$$
<p>

Where: 

* $R_{i,t}$: Expected return of the portfolio or stock.

* $R_f$: Risk-free rate.

* $R_{m,t}$: Expected market return. 

* $R_{m,t} - R_f$: *Equity market premium*

* $\beta_i$: The CAPM beta.

According to the CAPM, $\beta$ is the only relevant measure of risk, defined as:

<br>
$$
\beta_{i} = \frac{cov(R_{p,t}, R_{m,t})}{\sigma^{2}_m} 
$$
<br>

If $\beta$ is taken as the second coefficient of a simple OLS linear regression:

<br>
$$
R_{i,t} - R_{f} = \alpha_{i} + \beta_{i} (R_{m,t} - R_f) + \varepsilon_{i,t}
$$
<br>

Although the model above is not the CAPM, the regression allows the CAPM $\beta$ to be estimated. In fact, if $\alpha_i$ = 0 and $Cov(R_{m,t}, \varepsilon_{i,t}) = 0$ then:

<br>
$$
\begin{aligned}
R_{i,t} - R_{f} &= \beta_{i} (R_{m,t} - R_f) \\
R_{i,t}  &= R_{f} + \beta_{i} (R_{m,t} - R_f)
\end{aligned}
$$

The CAPM equation is obtained.
