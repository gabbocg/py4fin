3. Filter GameStop's OHCL data between 2020-01-01 and 2020-09-08. Then, using the `plotly` library, plot a *Candle-Stick* chart (see https://plotly.com/python/candlestick-charts/).

```
# imports plotly
import plotly.graph_objects as go

# filters the OHCL data
df_plotly = gme_ohcl_daily[(gme_ohcl_daily.index >= "2020-01-01") & (gme_ohcl_daily.index <= "2020-09-08")]

# plots candlestick
fig_plotly = go.Figure(data=[go.Candlestick(x=df_plotly.index,
                                            open=df_plotly['Open'],
                                            high=df_plotly['High'],
                                            low=df_plotly['Low'],
                                            close=df_plotly['Close'])])

# displays it
fig_plotly.show()
```
