7. The tangency portfolio is the one that maximizes the Sharpe ratio. The optimization problem expressed in its matrix form is:

    $$
    \max_{\mathbf{t}} \frac{\mathbf{t}'\mu - r_f}{(\mathbf{t}'\Sigma\mathbf{t})^{\frac{1}{2}}} = \frac{\mu_{p,t} - r_f}{\sigma_{p,t}} \quad s.t \quad \mathbf{t}'\mathbf{1} = 1
    $$

    Whose solution (also in matrix form) is:

    $$
    \mathbf{t} = \frac{\Sigma^{-1}(\mu - r_f \cdot \mathbf{1})}{\mathbf{1}'\Sigma^{-1}(\mu - r_f \cdot \mathbf{1})}
    $$

    Where $\mathbf{t}$ is the vector of optimal weights of the tangency portfolio, $\mathbf{1}$ a vector of ones whose dimension is $j \times 1$, with $j$ the total number of assets, and $\Sigma^{-1}$ the inverse of the variance-covariance matrix ($j \times j$).

        $$
        R_{p,t} = \mathbf{t}' \cdot \mu
        $$

        Where $\mu$ is the vector of average returns of dimension $j \times 1$.

    * The standard deviation of the minimum-variance portfolio is:

        $$
        \sigma_{p,t} = \sqrt{\mathbf{t}'\Sigma\mathbf{t}}
        $$
