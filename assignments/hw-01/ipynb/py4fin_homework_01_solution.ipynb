The file `returns_portfolios.pkl` contains a nested list that stores the monthly return of two portfolios ($j = 1, 2$) from 2019-01-01 to 2020-12-01, plus an annualized risk-free rate in percentage terms with the same frequency. Using this file, answer the following:

1. Import the file `returns_portfolios.pkl` as shown below:


````python
# imports the pandas library
import pandas as pd

# reads pickle
returns_portfolios = pd.read_pickle('returns_portfolios.pkl')
````

2. Compute the average return of each portfolio, defined as:

    $$
    r_j = \frac{\sum_{i=1}^{n} x_i}{n}
    $$


````python
total_ret_a = 0
total_ret_b = 0

for i, port in enumerate(returns_portfolios):
    for row in port:
        if i == 0:
            total_ret_a += row[1]
        else:
            total_ret_b += row[1]

mean_a = total_ret_a / len(returns_portfolios[0])
mean_b = total_ret_b / len(returns_portfolios[1])
````

3. Compute the standard deviation of each portfolio, defined as:

    $$
    \sigma_j = \sqrt{\frac{\sum_{i=1}^{n} (x_i - r_j)^2}{n-1}}
    $$


````python
desv_squared_port_a = 0
desv_squared_port_b = 0

for j, port in enumerate(returns_portfolios):
    for row in port:
        if j == 0:
            desv_squared_port_a += (row[1] - mean_a) ** 2
        else:
            desv_squared_port_b += (row[1] - mean_b) ** 2

std_a = (desv_squared_port_a / (len(returns_portfolios[0]) - 1)) ** (1/2)
std_b = (desv_squared_port_b / (len(returns_portfolios[1]) - 1)) ** (1/2)
````

4. Compute the average of the risk-free rate. Since the risk-free rate is annualized, you must divide it by 12 to make it monthly and be consistent with the monthly return of the portfolios.


````python
monthly_rfree = 0

for k, port in enumerate(returns_portfolios):
    for row in port:
        if k == 0:
            monthly_rfree += row[2] / 1200
        else:
            break

mean_rfree = monthly_rfree / len(returns_portfolios[0])
````

5. Compute the Sharpe ratio for each portfolio, defined as:

    $$
    SR_j = \frac{r_j - r_f}{\sigma_j}
    $$

    Where $r_j$ is the average return of portfolio $j$, $r_f$ the average of the risk-free rate not in percentage terms and $\sigma_j$ the standard deviation of portfolio j.


````python
sr_a = (mean_a - mean_rfree) / std_a
sr_b = (mean_b - mean_rfree) / std_b
````

6. Check whether the following statements are true:

    a. $SR_1$ is greater than $SR_2$.


````python
sr_a > sr_b
````

    b. $SR_1$ is less than $SR_2$.


````python
sr_a < sr_b
````

    c. The difference between $SR_2$ and $SR_1$ is positive.


````python
sr_b - sr_a > 0
````
