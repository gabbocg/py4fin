# Variables

A variable in Python is a space in your computer's RAM where you can store results and objects.
To create a variable we use an "assignment", that is, an expression with an equality. What is to the left of the equals sign will be the name of the variable where the value is stored, and what is to the right of the equals sign is the value to be stored (*Aplicaciones de Manejo de Datos, 2020*).

For example, let's create the variable `phone_number` and store the value `12345678`:

```python
# = assignment operator (not the only one)
phone_number = 12345678 # the symbol = is the assignment operator
phone_number
```

In python it is not necessary to end a line of code with ; (semicolons), but they do serve a purpose:



```python
a = 1; b = 2; c = 3
```

You can also assign values to several variables in a single line of code:

```python
a, b, c = 1, 2, 3
```

# Data Types (scalars)

1.  What are the data types in Python? How can I identify them and convert them into other data types? 

    There are four data types or scalars: numeric (**integers** and **floats**),   **strings** (text strings) and **booleans** (True & False).


```python
'''Basic data types'''
# integers
integer = 2 
type(integer) # identification
isinstance(integer, int)  # identification 

# float
floating = 3.14
type(floating)
isinstance(floating, float)

# string
string = "hello world"
type(string)
isinstance(string, str)

# boolean
boolean = True
type(boolean)
isinstance(boolean, bool)

'''casting'''
# integers
int('1')
int(True)
int(False)

# float
float('1')
float(True)
float(False)

# string
str(1)
str(True)
str(False)

# boolean
bool(1)
bool(0)
```

$$
PV = \sum^{T}_{t=0} \frac{F_t}{(1+r)^t}
$$


```python
present_value = 1000 / (1 + 0.1) ** 1 +  1000 / (1 + 0.1) ** 2 + 1000 / (1 + 0.1) ** 3 + 1000 / (1 + 0.1) ** 4 + 1000 / (1 + 0.1) ** 5
print('The present value is: ' + str(present_value))
```

3. Show that De Morgan's law holds.

| A     | B     | A and B | not(A and B)      | not A or not B                  |
|-------|-------|---------|-------------------|---------------------------------|
| True  | False | False   | not(False) = True | not(True) or not(False) = True  |
| False | True  | False   | not(False) = True | not(False) or not(True) = True  |
| True  | True  | True    | not(True) = False | not(True) or not(True) = False  |
| False | False | False   | not(False) = True | not(False) or not(False) = True |

```python
A = True
B = False

not(A and B) == (not(A) or not(B))
```

```python
A = False
B = True

not(A and B) == (not(A) or not(B))
```

```python
A = True
B = True

not(A and B) == (not(A) or not(B))
```

```python
A = False
B = False

not(A and B) == (not(A) or not(B))
```
